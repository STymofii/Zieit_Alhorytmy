## Завдання 2
**Приклади програм знаходження суми n чисел**

1. Через цикл
2. Через формулу


```python
from time import perf_counter

n = 10_000_000
# 1. Сума через цикл
def sum_cycle(n):
    total = 0
    for i in range(1, n+1):
        total += i
    start = perf_counter()
    return total, perf_counter()-start

start = perf_counter()
sum, t = sum_cycle(n)
print(f"Сума через цикл (n={n}): {sum}. Час виконання: {t}s")

# 2. Сума через формулу
def sum_formula(n):
    start = perf_counter()
    return n * (n + 1) // 2, perf_counter()-start


sum, t = sum_formula(n)
print(f"Сума через формулу (n={n}):{sum}. Час виконання: {t}s")
```

## Завдання 3
**Приклад алгоритму – Бінарний пошук**

```python
# Бінарний пошук (приклад)
def binary_search(lst, item):
    low = 0
    high = len(lst) - 1

    while low <= high:
        mid = (low + high) // 2
        guess = lst[mid]
        if guess == item:
            return mid
        if guess > item:
            high = mid - 1
        else:
            low = mid + 1
    return None

my_list = [1, 3, 5, 7, 9]
print("Пошук 3:", binary_search(my_list, 3))
print("Пошук 10:", binary_search(my_list, 10))
```
