## Завдання 2
**Приклади програм знаходження суми n чисел**

1. Через цикл
2. Через формулу


```python
from time import perf_counter

n = 10_000_000
# 1. Сума через цикл
def sum_cycle(n):
    start = perf_counter()
    total = 0
    for i in range(1, n+1):
        total += i
    return total, perf_counter()-start

start = perf_counter()
sum, t = sum_cycle(n)
print(f"Сума через цикл (n={n}): {sum}. Час виконання: {t}s")

# 2. Сума через формулу
def sum_formula(n):
    start = perf_counter()
    return n * (n + 1) // 2, perf_counter()-start


sum, t = sum_formula(n)
print(f"Сума через формулу (n={n}):{sum}. Час виконання: {t}s")
```

## Завдання 3
**Приклад алгоритму – Бінарний пошук**

```python
# Бінарний пошук (приклад)
def binary_search(lst, item):
    low = 0
    high = len(lst) - 1

    while low <= high:
        mid = (low + high) // 2
        guess = lst[mid]
        if guess == item:
            return mid
        if guess > item:
            high = mid - 1
        else:
            low = mid + 1
    return None

my_list = [1, 3, 5, 7, 9]
print("Пошук 3:", binary_search(my_list, 3))
print("Пошук 10:", binary_search(my_list, 10))
```

#### Ваша відповідь та обгрунтування тут:

1. Обчислення суми елементів масиву через цикл:

*    Часова складність: O(n)

*    Просторова складність: O(1)

*    Обгрунтування: Цикл виконується n разів (де n — кількість елементів масиву). На кожній ітерації здійснюється одна операція додавання, тому кількість дій зростає прямо пропорційно розміру масиву.

2. Обчислення суми елементів масиву через формулу:

*    Часова складність: O(1)

*    Просторова складність: O(1)

*    Обгрунтування: Використовується математична формула (наприклад, сума арифметичної прогресії S = n*(a1 + an)/2). Обчислення виконується за фіксовану кількість арифметичних операцій незалежно від величини n.

3. Лінійний пошук у списку:

*    Часова складність: O(n)

*    Просторова складність: O(1)

4. Бінарний пошук у списку:

*    Часова складність: O(log n)

*    Просторова складність: O(1)

## Контрольні запитання
1. Що таке алгоритм?

2. Чим визначається якість алгоритму?

3. Шо показує O-велике?

4. Який алгоритм більш ефективний з O(n) або з O(log n)? Чому?

1. Що таке алгоритм?

Це скінченна послідовність чітких правил і кроків, виконання яких призводить до розв'язання поставленої задачі за визначену кількість операцій.

2. Чим визначається якість алгоритму?

Якість алгоритму визначається такими основними критеріями:
*    Часова складність: швидкість виконання та кількість необхідних обчислювальних операцій.
*    Просторова складність: обсяг оперативної пам'яті, необхідний для роботи.
*    Коректність: здатність видавати правильний результат для всіх можливих валідних вхідних даних.
*    Масштабованість і читабельність: простота модифікації, підтримки та ефективність на великих обсягах даних.

3. Що показує O-велике?

Нотація O-велике O описує асимптотичну складність алгоритму — тобто те, як швидко зростає час виконання або обсяг пам'яті зі збільшенням розміру вхідних даних n, ігноруючи незначні константи та дрібні деталі реалізації.

4. Який алгоритм більш ефективний: з O(n) чи з O(log n)?

Більш ефективним є алгоритм з O(log n) (логарифмічна складність).
Чому: Логарифмічна залежність зростає набагато повільніше, ніж лінійна O(n).
