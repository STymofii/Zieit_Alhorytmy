````python
def Fact(n):
    if (n==0):
        return 1 # умова завершення рекурсивного процеса
    else:
        return n*Fact(n-1) # перехід до наступного числа n-1

print(Fact(5))
````

**Завдання для самостійної роботи:**

Реалізуйте алгоритм знаходження факторіалу з вводом числа n від користувача:



````python
def fact(n: int) -> int:
    if n < 0:
        raise ValueError("n має бути невід'ємним цілим числом")
    if n == 0:
        return 1
    return n * fact(n - 1)


n = int(input("Введіть ціле невід'ємне число n: "))
print(f"{n}! = {fact(n)}")
````

## Завдання 2
Напишіть НЕ рекурсивний алгоритм знаходження факторіалу числа n


````python
def fact_iter(n: int) -> int:
    if not isinstance(n, int) or isinstance(n, bool) or n < 0:
        raise ValueError("n має бути цілим невід'ємним числом")

    result = 1
    for i in range(2, n + 1):
        result *= i
    return result


n = int(input("Введіть ціле невід'ємне число n: "))
print(f"{n}! = {fact_iter(n)}")
````

## Завдання №3


Проведіть експерименти з вимірювання часу виконання обох алгоритмів для великих n та зробіть висновки.

````python
import sys
import time


def fact_rec(n: int) -> int:
    if n == 0:
        return 1
    return n * fact_rec(n - 1)


def fact_iter(n: int) -> int:
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result


sys.setrecursionlimit(200_000)

for n in [500, 1_000, 5_000, 10_000, 50_000]:
    t0 = time.perf_counter()
    try:
        fact_rec(n)
        t_rec = f"{time.perf_counter() - t0:.6f} с"
    except RecursionError:
        t_rec = "RecursionError"

    t0 = time.perf_counter()
    fact_iter(n)
    t_iter = f"{time.perf_counter() - t0:.6f} с"

    print(f"n = {n:6d} | Рекурсія: {t_rec:>14s} | Ітерація: {t_iter:>10s}")
````

10.Підрахунок кількості голосних у рядку
Рекурсивна функція, яка проходить рядок і рахує кількість голосних літер.

````python
def gcd(a: int, b: int) -> int:
    # Зводимо до невід'ємних значень: НСД(a, b) == НСД(|a|, |b|)
    a, b = abs(a), abs(b)

    # Базовий випадок рекурсії: коли остача 0, дільником є a
    if b == 0:
        return a

    # Рекурсивний крок: НСД(a, b) == НСД(b, a mod b)
    return gcd(b, a % b)


a = int(input("Введіть перше число: "))
b = int(input("Введіть друге число: "))
print(f"НСД({a}, {b}) = {gcd(a, b)}")
````

---


### 1. Рекурсія та її два елементи
* **Рекурсія:** Виклик функцією самої себе для розв'язання задачі через менші підзадачі того самого типу.
* **Базовий випадок:** Умова зупинки, яка повертає готовий результат без подальших викликів.
* **Рекурсивний крок:** Виклик функції зі зміненими (меншими) аргументами у напрямку до базового випадку.

### 2. Робота стека викликів
* Кожен виклик функції створює новий кадр стека з власними локальними змінними та точкою повернення і кладе його поверх попередніх.
* Після досягнення базового випадку кадри знімаються зі стека у зворотному порядку (LIFO), передаючи результати назад.

### 3. Потенційні проблеми рекурсії
* **Переповнення стека (`RecursionError`):** Занадто глибока рекурсія або відсутній/недосяжний базовий випадок.
* **Накладні витрати:** Додаткова пам'ять під кадри стека та час на виклики функцій порівняно з циклом.
* **Повторні обчислення:** Наївна рекурсія може багато разів рахувати одні й ті самі підзадачі.
