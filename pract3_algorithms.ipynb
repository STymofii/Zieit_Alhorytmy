````python
def selection_sort(array):
    length = len(array)
    # Перебираємо кожен елемент масиву (крім останнього)
    for i in range(length - 1):
        # Індекс найменшого елемента у поточній ітерації
        min_index = i

        # Шукаємо менший елемент у решті масиву
        for j in range(i + 1, length):
            if array[j] < array[min_index]:
                min_index = j

        # Міняємо місцями знайдений найменший з поточним
        array[i], array[min_index] = array[min_index], array[i]
    return(array)

arr = [5, 1, 9, 3, 7, 2, 8, 4, 10, 6]
print(selection_sort(arr))
````

````python
def bubble_sort(a):
    size = len(a)
    for i in range(size):
        for j in range(size - 1, i, -1):
            if a[j - 1] > a[j]:
                a[j - 1], a[j] = a[j], a[j - 1]
    return(a)


arr = [5, 1, 9, 3, 7, 2, 8, 4, 10, 6]
print(bubble_sort(arr))
````

## Завдання №3


Проведіть експерименти з вимірювання часу виконання обох алгоритмів для великих масивів та зробіть висновки.

````python
import random
import time

def bubble_sort_desc(arr):
    n = len(arr)
    for i in range(n - 1):
        swapped = False
        for j in range(n - 1 - i):
            if arr[j] < arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break

def selection_sort_desc(arr):
    n = len(arr)
    for i in range(n - 1):
        max_idx = i
        for j in range(i + 1, n):
            if arr[j] > arr[max_idx]:
                max_idx = j
        arr[i], arr[max_idx] = arr[max_idx], arr[i]

for size in [1000, 5000, 10000]:
    base_arr = [random.uniform(-1000, 1000) for _ in range(size)]

    arr1 = base_arr.copy()
    t0 = time.perf_counter()
    bubble_sort_desc(arr1)
    t_bubble = time.perf_counter() - t0

    arr2 = base_arr.copy()
    t0 = time.perf_counter()
    selection_sort_desc(arr2)
    t_select = time.perf_counter() - t0

    print(f"N = {size:5d} | Бульбашка: {t_bubble:.4f} с | Вибір: {t_select:.4f} с")
````

#### Результати експериментів та ваші висновки:

1. N =  1000 | Бульбашка: 0.1024 с | Вибір: 0.0491 с
2. N =  5000 | Бульбашка: 2.6044 с | Вибір: 0.6848 с
3. N = 10000 | Бульбашка: 6.7071 с | Вибір: 3.3823 с

Обидва алгоритми мають квадратичну складність O(N^2) і сильно сповільнюються зі зростанням масиву, але на випадкових даних сортування вибором приблизно удвічі-учетверичі швидше за бульбашку через мінімум обмінів у пам'яті. Натомість оптимізоване сортування бульбашкою з swapped виграє на вже відсортованих або майже відсортованих масивах, завершуючи роботу за O(N), тоді як сортування вибором усе одно виконує повний перебір.

## Завдання 4
Самостійно вивчить та наведіть реалізацію будь-якого іншого алгоритму сортування. Поясніть основні кроки в коментарях.

````python
def insertion_sort_desc(arr):
    # Починаємо з другого елемента (індекс 1), бо зріз [0] вже тривіально відсортований
    for i in range(1, len(arr)):
        key = arr[i]  # Зберігаємо поточний елемент, який потрібно вставити
        j = i - 1     # Індекс останнього елемента у відсортованій лівій частині

        # Зсуваємо вправо елементи, які менші за key (для порядку за спаданням)
        while j >= 0 and arr[j] < key:
            arr[j + 1] = arr[j]
            j -= 1

        # Ставимо key на звільнене місце
        arr[j + 1] = key

    return arr


arr = [5, 1, 9, 3, 7, 2, 8, 4, 10, 6]
print(insertion_sort_desc(arr))
````

### 1. Selection Sort
* **Ідея:** Шукає екстремум у невідсортованій частині й ставить його на початок цієї частини.
* **Складність:** O(N^2) за часом, O(1) за пам'яттю.

### 2. Bubble Sort vs Insertion Sort
* **Bubble Sort:** Міняє сусідні елементи, «виштовхуючи» крайні значення за прохід.
* **Insertion Sort:** Будує відсортований префікс і вставляє кожен наступний елемент у потрібне місце зі зсувом інших; зазвичай швидший за рахунок меншої кількості записів.

### 3. Найефективніший для великих даних
* **Timsort** (`sort()` / `sorted()` у Python): складність O(N log(N)) у середньому/найгіршому випадку та O(N) на частково відсортованих даних. Класичні O(N^2) для великих масивів не підходять.

### 4. Самостійний алгоритм (Insertion Sort) та відмінності
* **Алгоритм:** Сортування вставками (Insertion Sort).
* **Відмінності:** На відміну від Selection Sort, зупиняє внутрішній прохід раніше й працює за O(N) на вже впорядкованих даних; на відміну від Bubble Sort, зсуває елементи під збережений ключ замість постійних попарних обмінів.
